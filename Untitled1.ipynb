```python
# ============================================================
# StickerPro Ultimate v3.0
# セル① ライブラリ・基本設定
# ============================================================

!pip -q install pillow opencv-python

from google.colab import files
from PIL import Image
import numpy as np
import cv2
import os
import shutil

print("===================================")
print(" StickerPro Ultimate v3.0")
print("===================================")

# ----------------------------
# 分割設定
# ----------------------------
ROWS = 4
COLS = 4

# 出力サイズ
OUTPUT_SIZE = 512

# 出力フォルダ
OUTPUT_FOLDER = "line_stamp_png"

# 余白（LINE推奨）
MARGIN = 20

# ラベンダー背景色（あとで調整可能）
BG_COLOR = np.array([223, 204, 236])

# 色の許容範囲
BG_THRESHOLD = 35

print("初期化完了")
```

```python
# ============================================================
# StickerPro Ultimate v3.0
# セル② 画像アップロード・4×4分割
# ============================================================

print("4×4画像を選択してください")

uploaded = files.upload()

filename = next(iter(uploaded))

img = Image.open(filename).convert("RGBA")
img = np.array(img)

height, width = img.shape[:2]

cell_h = height // ROWS
cell_w = width // COLS

tiles = []

for row in range(ROWS):
    for col in range(COLS):

        y1 = row * cell_h
        y2 = (row + 1) * cell_h

        x1 = col * cell_w
        x2 = (col + 1) * cell_w

        tile = img[y1:y2, x1:x2]

        tiles.append(tile)

print(f"✅ 分割完了：{len(tiles)}枚")
```

```python
# ============================================================
# StickerPro Ultimate v4.0
# セル③（前半）
# OpenCV版 背景除去
# ============================================================

from PIL import Image
import numpy as np
import cv2

# ------------------------------------------------------------
# OpenCV背景除去
# ------------------------------------------------------------

def remove_background(img):

    img = img.convert("RGBA")

    arr = np.array(img)

    # RGB→HSV
    rgb = arr[:, :, :3]
    hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)

    # ラベンダー色を抽出
    lower = np.array([120, 15, 160], dtype=np.uint8)
    upper = np.array([170, 120, 255], dtype=np.uint8)

    mask = cv2.inRange(hsv, lower, upper)

    # ノイズ除去
    kernel = np.ones((3, 3), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)

    # 少しぼかして境界を自然に
    mask = cv2.GaussianBlur(mask, (5, 5), 0)

    alpha = 255 - mask

    arr[:, :, 3] = alpha

    return Image.fromarray(arr)


# ------------------------------------------------------------
# 自動トリミング
# ------------------------------------------------------------

def trim_image(img):

    alpha = img.getchannel("A")

    bbox = alpha.getbbox()

    if bbox is None:
        return img

    return img.crop(bbox)


# ------------------------------------------------------------
# 分割線除去
# ------------------------------------------------------------

def crop_border(img):

    w, h = img.size

    return img.crop((
        3,
        3,
        w - 3,
        h
    ))

print("✅ セル③ v4.0 前半 完成")
# ------------------------------------------------------------
# 512×512へ中央配置
# ------------------------------------------------------------

def fit_canvas(img, size=OUTPUT_SIZE, margin=MARGIN):

    img = trim_image(img)

    w, h = img.size

    if w == 0 or h == 0:
        return Image.new(
            "RGBA",
            (size, size),
            (0, 0, 0, 0)
        )

    scale = min(
        (size - margin * 2) / w,
        (size - margin * 2) / h
    )

    nw = max(1, int(w * scale))
    nh = max(1, int(h * scale))

    img = img.resize((nw, nh), Image.LANCZOS)

    canvas = Image.new(
        "RGBA",
        (size, size),
        (0, 0, 0, 0)
    )

    x = (size - nw) // 2
    y = (size - nh) // 2

    canvas.paste(img, (x, y), img)

    return canvas


# ------------------------------------------------------------
# 1枚を仕上げる
# ------------------------------------------------------------

def process_tile(tile):

    img = Image.fromarray(tile)

    # 分割線除去
    img = crop_border(img)

    # ラベンダー背景透過
    img = remove_background(img)

    # トリミング
    img = trim_image(img)

    # 512×512へ配置
    img = fit_canvas(img)

    return np.array(img)


print("✅ セル③ 完成")
```

```python
# ============================================================
# StickerPro Ultimate v4.0
# セル④ 一括画像処理
# ============================================================

output_tiles = []

print("===================================")
print("画像処理開始")
print("===================================")

for i, tile in enumerate(tiles, start=1):

    img = Image.fromarray(tile)

    # ------------------------------------------------
    # 分割線除去
    # ------------------------------------------------
    img = crop_border(img)

    # ------------------------------------------------
    # 左上の番号を削除
    # ------------------------------------------------
    img = img.crop((
        28,   # 左
        28,   # 上
        img.width,
        img.height
    ))

    # ------------------------------------------------
    # 背景透過
    # ------------------------------------------------
    img = remove_background(img)

    # ------------------------------------------------
    # トリミング
    # ------------------------------------------------
    img = trim_image(img)

    # ------------------------------------------------
    # 512×512へ配置
    # ------------------------------------------------
    img = fit_canvas(
        img,
        size=OUTPUT_SIZE,
        margin=MARGIN
    )

    output_tiles.append(np.array(img))

    print(f"✅ {i:02d}/16 完了")

print("===================================")
print(f"処理完了：{len(output_tiles)}枚")
print("===================================")
```

```python
# ============================================================
# StickerPro Ultimate v3.0
# セル⑤ PNG保存・ZIP作成・ダウンロード
# ============================================================

from PIL import Image
from google.colab import files
import os
import shutil

# 出力フォルダ作成
if os.path.exists(OUTPUT_FOLDER):
    shutil.rmtree(OUTPUT_FOLDER)

os.makedirs(OUTPUT_FOLDER)

print("===================================")
print("PNG保存開始")
print("===================================")

for i, tile in enumerate(output_tiles, start=1):

    img = Image.fromarray(tile)

    filename = os.path.join(
        OUTPUT_FOLDER,
        f"stamp_{i:02d}.png"
    )

    img.save(
        filename,
        "PNG",
        optimize=True
    )

print(f"{len(output_tiles)}枚 保存完了")

# ZIP作成
zip_name = OUTPUT_FOLDER + ".zip"

if os.path.exists(zip_name):
    os.remove(zip_name)

shutil.make_archive(
    OUTPUT_FOLDER,
    "zip",
    OUTPUT_FOLDER
)

print("ZIP作成完了")

# ダウンロード
files.download(zip_name)

print("===================================")
print("🎉 StickerPro Ultimate v3.0 完成！")
print("===================================")
```
